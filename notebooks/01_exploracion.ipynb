```python
#fijamos una semilla para que los resultados sean siempre iguales (reproducibles)
np.random.seed(42)

#1. lista de productos unicos
productos = basket["product_id"].unique()

#2. categorias y sus rangos de precios (USD)
categorias = {
    "Electronica": (50, 1500),
    "Ropa": (10, 200),
    "Hogar": (15, 500),
    "Belleza": (5, 120),
    "Deportes": (20, 400),
    "Libros": (5, 60)
}

nombres_cat = list(categorias.keys())

#3. asignar una categoria al azar a cada producto
cat_asignada = np.random.choice(nombres_cat, size=len(productos))

#4. generar un precio segun el rango de su categoria
precios = []
for cat in cat_asignada:
    minimo, maximo = categorias[cat]
    precio = round(np.random.uniform(minimo, maximo), 2)
    precios.append(precio)

#5. construir la tabla del catalogo
catalogo = pd.DataFrame({
    "product_id": productos,
    "category": cat_asignada,
    "price": precios
})

#6 guardar en un archivo CSV
catalogo.to_csv("product_catalog.csv", index=False)

#ver las primeras filas
print("catalogo creado con", len(catalogo), "productos")
catalogo.head()
```

```python
#unir basket con el catalogo usando producto_id como llave
ventas = basket.merge(catalogo, on="product_id", how="left")

#ver el resultado
print("Filas en ventas:", len(ventas))
ventas.head()
```

```python
#crear columna de ingreso: precio x cantidad comprada
ventas["revenue"] = ventas["price"] * ventas["basket_count"]

#ver el resultado
ventas.head()
```

```python
import sqlite3

#crear conexion a una base de datos SQLITE
conn = sqlite3.connect("ecommerce.db")

#subir muestras tablas a la base de datos
ventas.to_sql("ventas", conn, if_exists="replace", index=False)
customers.to_sql("customers", conn, if_exists="replace", index=False)

print("Base de datos lista. Ya podemos usar SQL")
```

```python
#funcion para ejecutar SQL y mostrar el resultado como tabla
def sql(query):
    return pd.read_sql_query(query, conn)
```

```python
sql(""" 
    WITH ranking AS (
        SELECT category,
                product_id,
                SUM(basket_count) AS unidades,
                RANK() OVER (PARTITION BY category ORDER BY SUM(basket_count) DESC) AS puesto
        FROM ventas
        GROUP BY category, product_id
    )
    SELECT category, product_id, unidades
    FROM ranking
    WHERE puesto = 1
""")
```

```python
sql(""" 
    SELECT strftime('%Y-%m', basket_date) AS mes,
            ROUND(SUM(revenue), 2) AS ingresos
    FROM ventas
    GROUP BY mes
    ORDER BY mes
""")
```

```python
sql(""" 
   WITH ventas_mes AS (
        SELECT strftime('%Y-%m', basket_date) AS mes,
               SUM(revenue) AS ingresos
        FROM ventas
        GROUP BY mes
    )
    SELECT mes,
           ROUND(ingresos, 2) AS ingresos,
           ROUND(LAG(ingresos) OVER (ORDER BY mes), 2) AS mes_anterior,
           ROUND(ingresos - LAG(ingresos) OVER (ORDER BY mes), 2) AS diferencia,
           ROUND((ingresos - LAG(ingresos) OVER (ORDER BY mes)) 
                 / LAG(ingresos) OVER (ORDER BY mes) * 100, 2) AS crecimiento_pct
    FROM ventas_mes
    ORDER BY mes
""")
```

```python
sql(""" 
    SELECT category,
            ROUND(SUM(revenue), 2) AS ingresos,
            ROUND(SUM(revenue) * 100.0 / (SELECT SUM(revenue) FROM ventas), 2) AS porcentaje
    FROM ventas
    GROUP BY category
    ORDER BY porcentaje DESC
""")
```

```python
sql("""
    WITH gasto_cliente AS (
        SELECT customer_id,
                SUM(revenue) AS total_gastado
        FROM ventas
        GROUP BY customer_id
    )
    SELECT customer_id,
            ROUND(total_gastado, 2) AS total_gastado,
            CASE
                WHEN total_gastado > 3000 THEN 'VIP'
                WHEN total_gastado >= 1000 THEN 'Frecuente'
                ELSE 'Ocacional'
            END AS tipo_cliente
    FROM gasto_cliente
    ORDER BY total_gastado DESC
    LIMIT 15
""")
```

```python
sql("""
    WITH gasto_cliente AS (
        SELECT customer_id,
                SUM(revenue) AS total_gastado
        FROM ventas
        GROUP BY customer_id
    )
    SELECT CASE
                WHEN total_gastado > 3000 THEN 'VIP'
                WHEN total_gastado >= 1000 THEN 'Frecuente'
                ELSE 'Ocasional'
            END AS tipo_cliente,
            COUNT(*) AS cantidad_clientes
    FROM gasto_cliente
    GROUP BY tipo_cliente
    ORDER BY cantidad_clientes DESC
""")
```

```python
sql("""
    WITH gasto_cliente AS (
        SELECT customer_id,
                SUM(revenue) AS total_gastado
        FROM ventas
        GROUP BY customer_id
    )
    SELECT customer_id,
            ROUND(total_gastado, 2) AS total_gastado
    FROM gasto_cliente
    WHERE total_gastado > (SELECT AVG(total_gastado) FROM gasto_cliente)
    ORDER BY total_gastado DESC
    LIMIT 15
""")
```

```python
#gasto total por cliente (desde VENTAS, igual que mi sql)
seg = ventas.groupby("customer_id")["revenue"].sum().reset_index()
seg.columns = ["customer_id", "gasto_total"]

#clasificar (mismo criterio que la pregunta 14 que respondi)
def clasificar(g):
    if g > 3000:
        return "VIP"
    elif g >= 1000:
        return "Frecuente"
    else:
        return "Ocacional"
seg["tipo_cliente"] = seg ["gasto_total"].apply(clasificar)

#exportar listo para power bi
seg.to_csv("segmentacion_cliente1.csv", index=False)

print("Archivo creado: segmentacio_clientes.csv")
print(seg["tipo_cliente"].value_counts())
print("\nTotal cliente:", len (seg))
```
